```python
# Cell 6 — flag implausible values based on domain logic (adjust thresholds after reviewing Cell 5)
implausible_rules = {
    "dti": lambda x: (x < 0) | (x > 100),
    "revol_util": lambda x: (x < 0) | (x > 150),   # can exceed 100% technically, but not by a huge margin
    "annual_inc": lambda x: x < 0,
    "open_acc": lambda x: x < 0,
}

flag_summary = {}
for col, rule in implausible_rules.items():
    flagged = rule(loan_df2[col])
    flag_summary[col] = flagged.sum()
    loan_df2.loc[flagged, col] = np.nan   # convert confirmed-impossible values to true missing, not a guessed number

pd.Series(flag_summary, name="n_flagged_and_nulled")
```

### IV/WOE test

```python
from optbinning import BinningProcess
```

```python
# Cell 4 — time-based split using issue_d (train only, per your OOT strategy)
loan_df2["issue_d"] = pd.to_datetime(loan_df2["issue_d"], format="%b-%Y")

train = loan_df2[loan_df2["issue_d"].dt.year <= 2016].copy()
val   = loan_df2[loan_df2["issue_d"].dt.year == 2017].copy()
test  = loan_df2[loan_df2["issue_d"].dt.year == 2018].copy()

train.shape, val.shape, test.shape
```

```python
# Cell 5 — leaked / post-origination columns to exclude before IV screening
# These describe what happened AFTER the loan was funded, or are identifiers/targets

leaked_cols = [
    # payment history / outcome of this loan
    "out_prncp", "out_prncp_inv", "total_pymnt", "total_pymnt_inv",
    "total_rec_prncp", "total_rec_int", "total_rec_late_fee",
    "recoveries", "collection_recovery_fee",
    "last_pymnt_d", "last_pymnt_amnt", "next_pymnt_d",
    "last_credit_pull_d", "last_fico_range_high", "last_fico_range_low",

    # target-derived / redundant target info
    "loan_status", "loan_status_clean",

    # hardship / settlement programs (only exist post-origination, post-distress)
    "hardship_flag", "hardship_type", "hardship_reason", "hardship_status",
    "deferral_term", "hardship_amount", "hardship_start_date", "hardship_end_date",
    "payment_plan_start_date", "hardship_length", "hardship_dpd",
    "hardship_loan_status", "orig_projected_additional_accrued_interest",
    "hardship_payoff_balance_amount", "hardship_last_payment_amount",
    "debt_settlement_flag", "debt_settlement_flag_date",
    "settlement_status", "settlement_date", "settlement_amount",
    "settlement_percentage", "settlement_term",

    # identifiers, free text, non-predictive
    "id", "member_id", "url", "desc", "emp_title", "title", "zip_code",

    # not a feature for modeling — used only for the time split
    "issue_d",

    # policy code — constant / not a real predictor
    "policy_code",
]

# drop only the ones actually present in this dataset version
leaked_cols_present = [c for c in leaked_cols if c in train.columns]
leaked_cols_present
```

```python
# Cell 6 — build the feature list: everything except target + leaked columns
exclude = leaked_cols_present + ["target"]
feature_cols = [c for c in train.columns if c not in exclude]

X_train = train[feature_cols]
y_train = train["target"].values

len(feature_cols)
```

```python
# Cell 7 — run IV/WOE screening with optbinning's BinningProcess
# This automatically handles missing values as their own bin, per feature type

# split numeric vs categorical for optbinning
numeric_cols = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = X_train.select_dtypes(include=["object"]).columns.tolist()
```

```python
# Cell 1 — find the offending columns
import numpy as np
import pandas as pd

diag = pd.DataFrame({
    "n_non_null": X_train.notna().sum(),
    "pct_missing": X_train.isna().mean().round(4),
    "n_unique": X_train.nunique(dropna=True),
})

# columns optbinning cannot fit at all
empty_cols = diag.index[diag["n_non_null"] == 0].tolist()

# columns that will fit but are effectively useless
constant_cols = diag.index[(diag["n_unique"] <= 1) & (diag["n_non_null"] > 0)].tolist()

print("100% missing in train:", empty_cols)
print("Constant in train:", constant_cols)
diag.sort_values("pct_missing", ascending=False).head(25)
```

```python
# Cell 2 — catch the sneakier cases (inf values, empty-string categoricals)
num_cols = X_train.select_dtypes(include=[np.number]).columns
all_inf = [c for c in num_cols
           if X_train[c].notna().sum() > 0
           and not np.isfinite(X_train[c].dropna()).any()]

obj_cols = X_train.select_dtypes(include=["object", "category"]).columns
blank_cols = [c for c in obj_cols
              if X_train[c].astype(str).str.strip()
                          .replace({"": np.nan, "nan": np.nan, "None": np.nan})
                          .notna().sum() == 0]

print("All-inf:", all_inf)
print("Blank-only categoricals:", blank_cols)
```

```python
# Cell 3 — drop and refit
drop_cols = sorted(set(empty_cols + constant_cols + all_inf + blank_cols))

feature_cols_clean = [c for c in feature_cols if c not in drop_cols]
categorical_cols_clean = [c for c in categorical_cols if c not in drop_cols]

print(f"Dropping {len(drop_cols)} columns, {len(feature_cols_clean)} remain")

binning_process = BinningProcess(
    variable_names=feature_cols_clean,
    categorical_variables=categorical_cols_clean,
)
binning_process.fit(X_train[feature_cols_clean], y_train)
```

```python
# Cell 7 — IV/WOE screening with optbinning's BinningProcess
# Fitted on the TRAINING SPLIT ONLY — no temporal leakage into feature selection.
# optbinning handles missing values as their own bin per feature.

# derive categoricals from the cleaned feature list, not from raw X_train
categorical_cols_clean = (
    X_train[feature_cols_clean]
    .select_dtypes(include=["object", "category", "bool"])
    .columns.tolist()
)

print(f"Features: {len(feature_cols_clean)}  |  Categorical: {len(categorical_cols_clean)}")

binning_process = BinningProcess(
    variable_names=feature_cols_clean,
    categorical_variables=categorical_cols_clean,
)

binning_process.fit(X_train[feature_cols_clean], y_train)
```
